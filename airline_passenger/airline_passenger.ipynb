```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('airline_passenger.csv')
```

```python
df.head()
```

```python
df = df.drop(columns='Unnamed: 0') # removing unimportant column
df.info()
```

```python
df.describe()
```

```python
print(f'The shape of the data: {df.shape}')
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
categorical_cols = ["Gender", "Customer Type", "Type of Travel",
                     "Class", "satisfaction"]

for col in categorical_cols:
    print(col, df[col].unique())
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(df['Flight Distance'], vert=False)

plt.title('Boxplot of Flight Distance')
plt.xlabel('Distance')

plt.show()
# the mathimatical outliers are logically valid as they can be a flight distance so will not be removed or replaced
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(df['Arrival Delay in Minutes'].dropna() , vert=False)

plt.title('Boxplot of Arrival Delay in Minutes')
plt.xlabel('Minutes')

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(df['Arrival Delay in Minutes'], bins=30, color='skyblue', edgecolor='black')
plt.title('Histogram of Column_Name')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

```python
mean_val = df['Arrival Delay in Minutes'].mean()
median_val = df['Arrival Delay in Minutes'].median()

print('The mean of the column: "Arrival Delay in Minutes" is: ', mean_val)
print('The median of the column: "Arrival Delay in Minutes" is: ', median_val)
```

```python
# replacing NaNs with median in column: 'Arrival Delay in Minutes' 
df['Arrival Delay in Minutes'] = df['Arrival Delay in Minutes'].fillna(median_val)
```

```python
df.isnull().sum()
```

```python
class_counts = df['Class'].value_counts()

fig, ax = plt.subplots(figsize=(5, 5))

colors = ['Green', 'Blue', 'red']

ax.pie(class_counts, labels=class_counts.index, autopct='%1.1f%%', startangle=90, 
       colors=colors, shadow=False, textprops={'fontsize': 12, 'weight': 'bold'})

ax.set_title('Passenger Distribution by Travel Class', fontsize=16, pad=20)

plt.show()
```

```python
class_satisfaction = df.groupby(['Class', 'satisfaction']).size().unstack()

fig, ax = plt.subplots(figsize=(8, 6))
class_satisfaction.plot(kind='bar', stacked=False, color=["red", "green"], ax=ax)

ax.set_title('Passenger Satisfaction by Travel Class', fontsize=14)
ax.set_xlabel('Travel Class', fontsize=12)
ax.set_ylabel('Number of Passengers', fontsize=12)
ax.legend(title='Satisfaction', loc='upper right')
plt.xticks(rotation=0)
plt.show()
```

```python
service_cols = [
    'Inflight wifi service', 'Departure/Arrival time convenient', 'Ease of Online booking', 
    'Gate location', 'Food and drink', 'Online boarding', 'Seat comfort', 
    'Inflight entertainment', 'On-board service', 'Leg room service', 
    'Baggage handling', 'Checkin service', 'Inflight service', 'Cleanliness'
]

mean_scores = df[service_cols].mean().sort_values(ascending=True)

fig, ax = plt.subplots(figsize=(10, 8))
bars = ax.barh(mean_scores.index, mean_scores.values, edgecolor='black', height=0.7)

ax.set_title('Airline Satisfaction Service Rates', fontsize=14, pad=15)
ax.set_xlabel('Average Rating (Out of 5)', fontsize=12)
ax.set_ylabel('Service Category', fontsize=12)
ax.set_xlim(0, 5)

for bar in bars:
    width = bar.get_width()
    ax.text(width + 0.05, bar.get_y() + bar.get_height()/2, f'{width:.2f}', 
            va='center', ha='left', fontsize=10)

plt.show()
```
